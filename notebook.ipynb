```python
from transformers import pipeline, BertTokenizer, BertForTokenClassification
from transformers import TokenClassificationPipeline

# Initialize the BERT tokenizer and model for NER (Named Entity Recognition)
tokenizer = BertTokenizer.from_pretrained('dbmdz/bert-large-cased-finetuned-conll03-english')
model = BertForTokenClassification.from_pretrained('dbmdz/bert-large-cased-finetuned-conll03-english')

# Set up a BERT NER pipeline
ner_pipeline = TokenClassificationPipeline(model=model, tokenizer=tokenizer)

def extract_skills_or_entities(text):
    """
    Use BERT to extract named entities (e.g., skills, qualifications) from the resume or job description.
    """
    ner_results = ner_pipeline(text)
    entities = [entity['word'] for entity in ner_results if entity['entity'] in ['ORG', 'MISC', 'PER', 'LOC']]
    return entities

# Initialize a text generation pipeline (e.g., GPT-Neo or any other model of your choice)
generator = pipeline('text-generation', model='EleutherAI/gpt-neo-125M')

def generate_cover_letter(resume_text, job_description):
    # Extract entities or skills from the resume and job description
    resume_entities = extract_skills_or_entities(resume_text)
    job_entities = extract_skills_or_entities(job_description)

    # Combine the extracted entities (e.g., skills, qualifications)
    combined_entities = list(set(resume_entities + job_entities))

    # Use the combined entities to form a customized prompt
    prompt = f"""
    Generate a personalized, ATS-friendly cover letter using the following entities:
    Resume: {resume_text}
    Job Description: {job_description}

    Extracted Entities/Skills: {', '.join(combined_entities)}

    Cover Letter:
    """
    
    # Customize parameters like max_length, temperature, etc., as needed
    generated = generator(prompt, max_new_tokens=200, num_return_sequences=1, do_sample=True, temperature=0.7)
    
    return generated[0]['generated_text']

# Example usage
resume_text = "I am a software engineer with skills in Python, Java, and machine learning."
job_description = "Looking for a software engineer proficient in Python, JavaScript, and AI."

cover_letter = generate_cover_letter(resume_text, job_description)
print(cover_letter)
```
